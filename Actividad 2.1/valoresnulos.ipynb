```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('Citas_Digital_Filtrado.csv')
data.head(5)
```

```python
#Identificar valores nulos por columna
valores_nulos = data.isnull().sum()
valores_nulos
```

```python
#Identificar valores nulos en todo el dataframe
valores_nulos = data.isnull().sum().sum()
valores_nulos
```

```python
## Eliminar columnas que no aportan
#La columna 'Unnamed: 15' es un residuo del archivo original (sin nombre y casi vacia), no aporta datos
#axis = 1 para columnas
data1 = data.drop(['Unnamed: 15'], axis=1)
data1
```

```python
#Corroboramos valores nulos
valores_nulos = data1.isnull().sum()
valores_nulos
```

```python
##Metodos de sustitución de valores nulos
#Realizamos una copia del dataframe
data2 = data1.copy()
```

```python
#Primer método de sustitución de valores nulos
#Sustituir valores nulos por un string en concreto (columna categorica con pocos nulos)
data2["BDC "] = data2["BDC "].fillna("Sin dato")
```

```python
#Segundo método de sustitución de valores nulos
#Sustituir valores nulos por un string en concreto
data2["Telefono "] = data2["Telefono "].fillna("Sin telefono")
```

```python
#Tercer método de sustitución de valores nulos
#Sustituir valores nulos por un string en concreto 
data2["Fecha"] = data2["Fecha"].fillna("Sin fecha registrada")
```

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en concreto (no se puede inventar el nombre del cliente)
data2["Nombre Cliente "] = data2["Nombre Cliente "].fillna("Desconocido")
```

```python
#Quinto método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia adelante "forward fill" ("ffill")
#Filtro por columnas
data2["Estatus de Lead"] = data2["Estatus de Lead"].ffill()
```

```python
#Sexto método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia atrás "backward fill" ("bfill")
#Filtro por columnas
data2["Asesor Asignado"] = data2["Asesor Asignado"].bfill()
```

```python
#Septimo método de sustitución de valores nulos
#La columna tiene datos sucios (texto suelto y simbolos de porcentaje mal escritos), los limpiamos primero
data2["Potencial de compra"] = data2["Potencial de compra"].replace({
    "Ventas ": np.nan,
    "80%%": "0.8",
    "8'%": "0.8",
    "80&": "0.8"
})

#Convertimos la columna a numerica (venia como texto) y sustituimos los nulos con el promedio
data2["Potencial de compra"] = data2["Potencial de compra"].astype(float)
data2["Potencial de compra"] = data2["Potencial de compra"].fillna(round(data2["Potencial de compra"].mean(), 2))
```

```python
#Octavo método de sustitución de valores nulos
#Sustituir valores nulos por un valor numerico en concreto (PDM, SDC y Venta son columnas de 0 y 1)
data2["PDM"] = data2["PDM"].fillna(0)
data2["SDC"] = data2["SDC"].fillna(0)
data2["Venta"] = data2["Venta"].fillna(0)
```

```python
#Noveno método de sustitución de valores nulos
#Sustituir valores nulos por un string en concreto ("No aplica" ya que el nulo significa que esa pregunta no correspondia al cliente)
preguntas = ['¿Por que no ha visitado la agencia?',
             '¿Por que no hay solicitud de credito?',
             '¿Por que no hubo prueba de manejo ?',
             '¿Por que no hubo proceso wow?',
             '¿Por que se asigno antes de vistar piso?']

data2[preguntas] = data2[preguntas].fillna("No aplica")
```

```python
#Corroboramos valores nulos en todo el dataframe
valores_nulos = data2.isnull().sum().sum()
valores_nulos
```
